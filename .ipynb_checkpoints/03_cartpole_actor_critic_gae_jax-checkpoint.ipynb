# 03 — Actor–Critic and Generalized Advantage Estimation (GAE) in JAX

**Environment:** CartPole-v1 (discrete actions) · **Stack:** JAX, Flax, Optax, Gymnax.

We build a complete on-policy actor–critic agent with fixed-length batched rollouts, bootstrapped value targets, generalized advantage estimation, and reproducible evaluation. Experiments compare **TD(0), GAE($\lambda=0.5$), GAE($\lambda=0.95$), and Monte Carlo-like GAE($\lambda=1$)**.

This notebook is a bridge from Notebook 02 (REINFORCE) to Notebook 04 (PPO). Unlike REINFORCE with a Monte Carlo value baseline, **our critic bootstraps**, and the policy uses bootstrapped advantage estimates. Policy updates are *unclipped*; PPO is deliberately left for the next notebook.

**Runtime:** CPU-compatible. Execute all cells in order. The optional multi-seed sweep takes longer.


```python
# Uncomment if dependencies are missing. Restart the kernel after installation if necessary.
# %pip install -q "jax>=0.4.30" "flax>=0.10" "optax>=0.2.3" "gymnax>=0.0.8" matplotlib pandas
```

## 1. Derivation: TD error, GAE, and bootstrapping

Let $V_\phi(s)$ be the learned state-value function. For a nonterminal transition,

$\delta_t = r_t + \gamma V_\phi(s_{t+1}) - V_\phi(s_t)$.

For a terminal transition, the bootstrap term is removed:

$\delta_t = r_t + \gamma (1-d_t)V_\phi(s_{t+1}) - V_\phi(s_t)$.

Generalized advantage estimation is the backward recursion

$\hat A_t^{\gamma,\lambda} = \delta_t + \gamma\lambda(1-d_t)\hat A_{t+1}^{\gamma,\lambda}$.

We fit the critic to the **lambda-return target**

$\hat V_t^{\mathrm{target}} = \operatorname{stopgrad}(\hat A_t^{\gamma,\lambda} + V_\phi(s_t))$.

The actor loss is

$L_\pi = -\mathbb E_{t}[\log\pi_\theta(a_t\mid s_t)\operatorname{stopgrad}(\hat A_t)] - \beta\mathbb E_t[\mathcal H(\pi_\theta(\cdot\mid s_t))]$.

- $\lambda=0$: one-step TD advantage.
- $0<\lambda<1$: exponentially weighted multi-step TD errors.
- $\lambda=1$: Monte Carlo return minus baseline **up to termination**, but a truncated rollout still bootstraps from its final state.
- $\gamma$ controls discounting, while $\lambda$ controls the bias–variance trade-off of the advantage estimator.

**Truncation convention.** We use fixed-length rollout segments. When a segment ends while the environment is still alive, its final value is bootstrapped; true terminations are never bootstrapped. Gymnax CartPole has a time limit; a time-limit `done` is treated as terminal for this episodic experiment.


```python
import jax
import jax.numpy as jnp
import flax.linen as nn
import optax
import gymnax
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from dataclasses import dataclass, replace
from functools import partial
from jax.flatten_util import ravel_pytree

print("JAX", jax.__version__, "| devices", jax.devices())
env, env_params = gymnax.make("CartPole-v1")
```

## 2. Configuration and networks

```python
@dataclass(frozen=True)
class Config:
    seed: int = 0
    num_envs: int = 32
    rollout_steps: int = 64
    updates: int = 500
    gamma: float = 0.99
    gae_lambda: float = 0.95
    lr: float = 3e-4
    hidden: int = 64
    entropy_coef: float = 0.01
    value_coef: float = 0.5
    max_grad_norm: float = 0.5
    eval_episodes: int = 100
    max_episode_steps: int = 500

CFG = Config()

class ActorCritic(nn.Module):
    n_actions: int
    hidden: int = 64

    @nn.compact
    def __call__(self, obs):
        x = nn.tanh(nn.Dense(self.hidden)(obs))
        x = nn.tanh(nn.Dense(self.hidden)(x))
        return nn.Dense(self.n_actions, name="policy_head")(x), nn.Dense(1, name="value_head")(x)[..., 0]

# Infer dimensions rather than assuming a particular Gymnax version.
key = jax.random.PRNGKey(0)
sample_obs, sample_state = env.reset(key, env_params)
obs_dim = int(sample_obs.shape[-1])
n_actions = int(env.num_actions)
model = ActorCritic(n_actions, CFG.hidden)
print("Observation dimension:", obs_dim, "| actions:", n_actions)

def init_train_state(seed, cfg=CFG):
    params = model.init(jax.random.PRNGKey(seed), jnp.zeros((obs_dim,), jnp.float32))
    tx = optax.chain(optax.clip_by_global_norm(cfg.max_grad_norm), optax.adam(cfg.lr))
    return params, tx.init(params), tx

def env_step(key, state, action):
    # Gymnax variants may return additional values after (obs, state, reward, done).
    result = env.step(key, state, action, env_params)
    obs, next_state, reward, done = result[:4]
    return obs, next_state, reward, done
```

## 3. Vectorized rollouts with automatic reset

Unlike Notebook 02's fixed-length *episodic* scan, training here uses a **continuing stream of independent episodes**. Each of the 32 environments automatically resets after a terminal transition, so the next rollout step starts a fresh episode.

For GAE, each transition records its own `done` flag. This prevents bootstrapping across an episode boundary. We also retain the actual transition's successor value before resetting; the reset observation is used only for the *next* action. We store the next value of the **pre-reset** state, so a reset cannot accidentally be used as the bootstrap state.

Each update collects `num_envs × rollout_steps` fresh on-policy transitions.


```python
def reset_batch(keys):
    return jax.vmap(lambda k: env.reset(k, env_params))(keys)

def collect_rollout(params, obs, states, rng, num_steps):
    def scan_step(carry, _):
        obs, states, rng = carry
        rng, action_key, step_key, reset_key = jax.random.split(rng, 4)
        logits, values = jax.vmap(lambda x: model.apply(params, x))(obs)
        action_keys = jax.random.split(action_key, obs.shape[0])
        actions = jax.vmap(jax.random.categorical)(action_keys, logits)
        step_keys = jax.random.split(step_key, obs.shape[0])
        next_obs, next_states, rewards, dones = jax.vmap(env_step)(
            step_keys, states, actions
        )
        # Compute successor values *before* replacing terminal states with resets.
        _, next_values = jax.vmap(lambda x: model.apply(params, x))(next_obs)
        reset_keys = jax.random.split(reset_key, obs.shape[0])
        reset_obs, reset_states = reset_batch(reset_keys)
        carry_obs = jnp.where(dones[:, None], reset_obs, next_obs)
        carry_states = jax.tree_util.tree_map(
            lambda new, reset: jnp.where(
                jnp.reshape(dones, dones.shape + (1,) * (new.ndim - dones.ndim)),
                reset, new
            ), next_states, reset_states
        )
        # Store observations/actions from the *old* policy, rewards, and flags.
        data = (obs, actions, rewards, dones, values, next_values)
        return (carry_obs, carry_states, rng), data

    (obs, states, rng), data = jax.lax.scan(
        scan_step, (obs, states, rng), xs=None, length=num_steps
    )
    return obs, states, rng, data

collect_rollout_jit = jax.jit(collect_rollout, static_argnames=("num_steps",))

def fresh_envs(seed, n):
    rng = jax.random.PRNGKey(seed)
    rng, sub = jax.random.split(rng)
    obs, states = reset_batch(jax.random.split(sub, n))
    return obs, states, rng
```

## 4. GAE implementation and mathematical checks

The time-major arrays have shape `[T, N]`. We reverse-scan the TD residuals and stop propagation at episode boundaries. The lambda-return target is computed from **old, pre-update** value predictions and treated as constant during optimization.


```python
@jax.jit
def gae_advantages(rewards, dones, values, next_values, gamma, lam):
    nonterminal = 1.0 - dones.astype(rewards.dtype)
    deltas = rewards + gamma * nonterminal * next_values - values

    def backward(last_gae, xs):
        delta, live = xs
        advantage = delta + gamma * lam * live * last_gae
        return advantage, advantage

    _, reversed_adv = jax.lax.scan(
        backward, jnp.zeros_like(deltas[0]),
        (deltas[::-1], nonterminal[::-1])
    )
    adv = reversed_adv[::-1]
    targets = adv + values
    return adv, targets

# Analytic tests: gamma=1, values=0 => undiscounted returns for lambda=1.
r = jnp.array([[1.0], [2.0], [3.0]])
d = jnp.array([[False], [False], [True]])
v = jnp.zeros_like(r)
a1, target1 = gae_advantages(r, d, v, v, 1.0, 1.0)
assert np.allclose(np.asarray(a1[:, 0]), [6., 5., 3.])
a0, _ = gae_advantages(r, d, v, v, 1.0, 0.0)
assert np.allclose(np.asarray(a0[:, 0]), [1., 2., 3.])
# Check a terminal transition does not use the next state's value.
a_term, _ = gae_advantages(
    jnp.array([[2.0]]), jnp.array([[True]]),
    jnp.array([[0.5]]), jnp.array([[999.0]]), 0.99, 0.95
)
assert np.allclose(np.asarray(a_term), [[1.5]])
# Check segment truncation (not done) *does* bootstrap.
a_trunc, _ = gae_advantages(
    jnp.array([[2.0]]), jnp.array([[False]]),
    jnp.array([[0.5]]), jnp.array([[10.0]]), 0.99, 0.95
)
assert np.allclose(np.asarray(a_trunc), [[11.4]])
print("GAE terminal/truncation and lambda-limit tests passed.")
```

## 5. One on-policy actor–critic update

We minimize a combined actor, critic, and entropy objective. This notebook intentionally uses **one gradient update per rollout** (no minibatch epochs, no importance ratios). Thus the data remain on-policy at every update. Advantage normalization is optional in general; here we use it to improve numerical stability. Because normalization rescales the actor gradient, it changes the effective optimization dynamics without changing the relative sign of advantages.

The value target is stopped-gradient; the critic is updated using the same fresh rollout as the actor.


```python
def make_update(cfg, tx):
    def loss_fn(params, data, adv, targets):
        obs, actions, rewards, dones, old_values, next_values = data
        T, N = actions.shape
        flat_obs = obs.reshape((T * N, obs.shape[-1]))
        logits, values = model.apply(params, flat_obs)
        flat_actions = actions.reshape(-1)
        selected_logprob = jnp.take_along_axis(
            jax.nn.log_softmax(logits), flat_actions[:, None], axis=-1
        )[:, 0]
        entropy = -jnp.sum(
            jax.nn.softmax(logits) * jax.nn.log_softmax(logits), axis=-1
        ).mean()
        flat_adv = jax.lax.stop_gradient(adv.reshape(-1))
        norm_adv = (flat_adv - flat_adv.mean()) / (flat_adv.std() + 1e-8)
        actor_loss = -(selected_logprob * norm_adv).mean()
        critic_loss = 0.5 * jnp.mean(
            (values - jax.lax.stop_gradient(targets.reshape(-1))) ** 2
        )
        total = actor_loss + cfg.value_coef * critic_loss - cfg.entropy_coef * entropy
        return total, (actor_loss, critic_loss, entropy)

    @jax.jit
    def update(params, opt_state, data):
        obs, actions, rewards, dones, values, next_values = data
        adv, targets = gae_advantages(
            rewards, dones, values, next_values, cfg.gamma, cfg.gae_lambda
        )
        (loss, (al, vl, entropy)), grads = jax.value_and_grad(
            loss_fn, has_aux=True
        )(params, data, adv, targets)
        updates, opt_state = tx.update(grads, opt_state, params)
        params = optax.apply_updates(params, updates)
        return params, opt_state, (loss, al, vl, entropy, jnp.mean(rewards),
                                    jnp.mean(dones.astype(jnp.float32)))
    return update

# tx is a Python optimizer object and must be closed over, not passed as a traced argument.
def make_compiled_update(cfg, tx):
    raw_update = make_update(cfg, tx)
    @jax.jit
    def update(params, opt_state, data):
        return raw_update(params, opt_state, data)
    return update
```

## 6. Training and sanity checks

```python
def train_actor_critic(cfg=CFG, seed=None, verbose=True):
    seed = cfg.seed if seed is None else seed
    params, opt_state, tx = init_train_state(seed, cfg)
    update_fn = make_compiled_update(cfg, tx)
    obs, states, rng = fresh_envs(seed + 100, cfg.num_envs)
    history = []
    for iteration in range(cfg.updates):
        obs, states, rng, data = collect_rollout_jit(
            params, obs, states, rng, cfg.rollout_steps
        )
        params, opt_state, stats = update_fn(params, opt_state, data)
        loss, actor, critic, entropy, reward, terminal_fraction = map(float, stats)
        history.append({
            "update": iteration + 1,
            "environment_steps": (iteration + 1) * cfg.num_envs * cfg.rollout_steps,
            "loss": loss, "actor_loss": actor, "critic_loss": critic,
            "entropy": entropy, "reward_per_step": reward,
            "terminal_fraction": terminal_fraction
        })
        if verbose and (iteration + 1) % 100 == 0:
            print(f"lambda={cfg.gae_lambda:.2f} | update {iteration+1:4d} "
                  f"| reward/step {reward:.3f} | value loss {critic:.3f}")
    return params, pd.DataFrame(history)

# Fast end-to-end smoke test, including the environment interface and optimizer.
smoke_cfg = replace(CFG, num_envs=4, rollout_steps=8, updates=2)
smoke_params, smoke_history = train_actor_critic(smoke_cfg, verbose=False)
assert np.isfinite(smoke_history.select_dtypes(include="number").to_numpy()).all()
print("Rollout and optimizer smoke test passed.")
```

## 7. Evaluation with common random keys

For evaluation we run complete episodes (up to 500 steps), mask all post-terminal rewards, and freeze the observation/state after termination. Both **sampled** and **greedy** policies are evaluated, because the training objective concerns the sampled policy. All variants use identical evaluation reset keys.


```python
def eval_one(params, key, max_steps, greedy=False):
    rng, reset_key = jax.random.split(key)
    obs, state = env.reset(reset_key, env_params)
    def step(carry, _):
        obs, state, rng, done = carry
        rng, act_key, env_key = jax.random.split(rng, 3)
        logits, _ = model.apply(params, obs)
        action = (jnp.argmax(logits) if greedy
                  else jax.random.categorical(act_key, logits))
        new_obs, new_state, reward, step_done = env_step(env_key, state, action)
        reward = jnp.where(done, 0.0, reward)
        new_obs = jnp.where(done, obs, new_obs)
        new_state = jax.tree_util.tree_map(
            lambda old, new: jnp.where(done, old, new), state, new_state
        )
        return (new_obs, new_state, rng, done | step_done), reward
    _, rewards = jax.lax.scan(
        step, (obs, state, rng, jnp.bool_(False)), None, length=max_steps
    )
    return rewards.sum()

@partial(jax.jit, static_argnames=("max_steps", "greedy"))
def eval_batch(params, keys, max_steps=500, greedy=False):
    return jax.vmap(lambda k: eval_one(params, k, max_steps, greedy))(keys)

def evaluate(params, keys, cfg=CFG, greedy=False):
    return np.asarray(eval_batch(params, keys, cfg.max_episode_steps, greedy))

def bootstrap_ci(x, seed=0, n_boot=2000):
    x = np.asarray(x)
    rng = np.random.default_rng(seed)
    means = rng.choice(x, size=(n_boot, len(x)), replace=True).mean(axis=1)
    return np.quantile(means, [0.025, 0.975])
```

## 8. Compare GAE lambda values

```python
LAMBDAS = (0.0, 0.5, 0.95, 1.0)
runs = {}
for lam in LAMBDAS:
    print(f"\nTraining lambda={lam}")
    runs[lam] = train_actor_critic(replace(CFG, gae_lambda=lam), seed=CFG.seed)

fig, axes = plt.subplots(1, 3, figsize=(16, 4))
for lam, (_, hist) in runs.items():
    x = hist["environment_steps"]
    for ax, metric in zip(axes, ["reward_per_step", "critic_loss", "entropy"]):
        ax.plot(x, hist[metric].rolling(20, min_periods=1).mean(),
                label=f"λ={lam:g}")
for ax, title in zip(axes, ["Reward per collected step", "Critic MSE / 2",
                            "Policy entropy"]):
    ax.set(title=title, xlabel="Environment steps")
    ax.grid(alpha=.25)
axes[0].legend()
plt.tight_layout()
plt.show()

# Note: CartPole rewards are typically +1 until termination, so reward/step
# can saturate at 1 even when episode lengths are short. Evaluation return
# is the meaningful primary performance measure.
```

## 9. Final episodic evaluation

```python
eval_keys = jax.random.split(jax.random.PRNGKey(202603), CFG.eval_episodes)
scores = {}
summary_rows = []
for lam, (params, _) in runs.items():
    for mode in ("sampled", "greedy"):
        returns = evaluate(params, eval_keys, greedy=(mode == "greedy"))
        scores[(lam, mode)] = returns
        lo, hi = bootstrap_ci(returns)
        summary_rows.append({
            "lambda": lam, "mode": mode,
            "mean": returns.mean(), "median": np.median(returns),
            "std": returns.std(ddof=1), "CI_low": lo, "CI_high": hi,
        })
summary = pd.DataFrame(summary_rows)
display(summary.round(2))

fig, ax = plt.subplots(figsize=(9, 4))
ax.boxplot([scores[(lam, "sampled")] for lam in LAMBDAS],
           tick_labels=[f"λ={lam:g}" for lam in LAMBDAS], showmeans=True)
ax.set(title="Sampled-policy evaluation: paired initial conditions",
       ylabel="Undiscounted episode return")
ax.grid(axis="y", alpha=.25)
plt.tight_layout()
plt.show()
```

## 10. Advantage estimator variability at fixed parameters

To isolate the estimator from differences in learned policies, we collect fresh rollouts using a single frozen network and compute the empirical variance of the **batch-mean advantages** and of the **un-normalized policy-gradient vectors** across batches for each lambda. These are diagnostics, not guarantees that a particular lambda will train best.

A fixed policy is used to generate all batches. Gradient computation uses the raw advantages without advantage normalization so that the diagnostic reflects the estimators rather than the normalization procedure.


```python
def policy_gradient_vector(params, data, advantages):
    obs, actions, *_ = data
    T, N = actions.shape
    def loss(p):
        logits, _ = model.apply(p, obs.reshape(T * N, -1))
        logp = jax.nn.log_softmax(logits)
        selected = jnp.take_along_axis(
            logp, actions.reshape(-1, 1), axis=-1
        )[:, 0]
        return -(selected * jax.lax.stop_gradient(advantages.reshape(-1))).mean()
    grads = jax.grad(loss)(params)
    return ravel_pytree(grads)[0]

def gradient_variance_experiment(reference_params, cfg=CFG, n_batches=20):
    obs, states, rng = fresh_envs(999, cfg.num_envs)
    grad_vectors = {lam: [] for lam in LAMBDAS}
    for _ in range(n_batches):
        obs, states, rng, data = collect_rollout_jit(
            reference_params, obs, states, rng, cfg.rollout_steps
        )
        _, _, rewards, dones, values, next_values = data
        for lam in LAMBDAS:
            adv, _ = gae_advantages(
                rewards, dones, values, next_values, cfg.gamma, lam
            )
            grad_vectors[lam].append(np.asarray(
                policy_gradient_vector(reference_params, data, adv)
            ))
    rows = []
    for lam, vectors in grad_vectors.items():
        arr = np.stack(vectors)
        rows.append({
            "lambda": lam,
            "gradient_cov_trace": np.var(arr, axis=0, ddof=1).sum(),
            "mean_gradient_norm": np.linalg.norm(arr.mean(axis=0))
        })
    return pd.DataFrame(rows)

gradient_stats = gradient_variance_experiment(runs[0.95][0])
display(gradient_stats.round(4))
ax = gradient_stats.plot.bar(x="lambda", y="gradient_cov_trace",
                             legend=False, figsize=(7, 4))
ax.set(ylabel="Trace of gradient covariance",
       title="Gradient variability across rollout batches")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 11. Multi-seed comparison (optional, recommended)

Run with `RUN_MULTISEED=True` for a stronger comparison. We aggregate at the **training-seed** level, not by treating hundreds of evaluation episodes as independent trained agents. The confidence intervals above quantify episode-level uncertainty conditional on each fitted policy; multi-seed variation captures sensitivity to optimization and data collection.


```python
RUN_MULTISEED = False
TRAIN_SEEDS = (0, 1, 2, 3, 4)

if RUN_MULTISEED:
    rows = []
    for seed in TRAIN_SEEDS:
        keys = jax.random.split(jax.random.PRNGKey(10000 + seed),
                                CFG.eval_episodes)
        for lam in LAMBDAS:
            params, history = train_actor_critic(
                replace(CFG, gae_lambda=lam), seed=seed, verbose=False
            )
            returns = evaluate(params, keys)
            rows.append({"seed": seed, "lambda": lam,
                         "mean_return": float(returns.mean()),
                         "median_return": float(np.median(returns))})
    multiseed = pd.DataFrame(rows)
    display(multiseed.pivot(index="seed", columns="lambda",
                            values="mean_return").round(1))
    display(multiseed.groupby("lambda")["mean_return"]
            .agg(["mean", "std"]).round(2))
```

## 12. Interpretation and limitations

1. **Bias–variance:** Small lambda uses more bootstrapping and may have lower sampling variance but higher bias from an inaccurate critic. Lambda near one uses longer-horizon information.
2. **Terminations:** GAE recursion stops at true episode boundaries. A fixed-length segment boundary is *not* terminal; its final value is used to bootstrap.
3. **Sampled vs greedy:** The actor is trained with sampled actions. Greedy evaluation is an additional diagnostic, not the objective optimized during training.
4. **No PPO yet:** This notebook performs a single on-policy gradient update per rollout; no clipped importance ratio, no multiple policy epochs.
5. **Interpretation of metrics:** Episode return is the primary metric. Per-step reward is often uninformative for CartPole because nearly every nonterminal step pays +1.
6. **Implementation scope:** One shared encoder feeds actor and critic. This is a practical actor–critic architecture but makes gradient variance comparisons reflect a shared parameterization.
